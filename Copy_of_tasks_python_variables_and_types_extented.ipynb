**Правила імен:** латинські літери, цифри та `_`; не можна починати з цифри; великі й малі літери різні (`Age` ≠ `age`). Зазвичай пишуть `snake_case`: `my_city`.

Щоб побачити значення, використовуємо функцію `print()`:

```python
print(fruit)
print(count)
```

## Розділ 2. Основні типи даних
**Мета:** попрактикуватись із `int`, `float`, `str`, `bool`.

> 💡 **Трохи теорії**

Основні базові типи:

| Тип | Що це | Приклад |
|---|---|---|
| `int` | ціле число | `7`, `-3`, `0` |
| `float` | дробове число (через **крапку**!) | `2.5`, `-0.1` |
| `str` | рядок, текст у лапках | `"Hello"`, `'Київ'` |
| `bool` | істина / хиба | `True`, `False` |

Дробову частину відокремлюємо крапкою: `3.5`, а не `3,5` (кома в Python означає зовсім інше).

> 💡 **Трохи теорії**

Арифметичні оператори:

| Оператор | Дія | Приклад | Результат |
|---|---|---|---|
| `+` `-` `*` | додавання, віднімання, множення | `4 * 3` | `12` |
| `/` | ділення (завжди дає `float`) | `7 / 2` | `3.5` |
| `//` | ціла частина від ділення | `7 // 2` | `3` |
| `%` | остача від ділення | `7 % 2` | `1` |
| `**` | піднесення до степеня | `2 ** 3` | `8` |

Якщо змішати `int` і `float`, результат буде `float`: `2 + 1.5` → `3.5`.

| `float(x)` | у дробове | `float("2.5")` → `2.5` |

Результат перетворення треба зберегти в змінну: `a = str(a)`.

## Розділ 5. Логічні вирази (`bool`)

> 💡 **Трохи теорії**

Результат порівняння можна **зберегти у змінну** — її тип буде `bool`:

```python
temperature = 30
is_hot = temperature > 25
print(is_hot)        # True
print(type(is_hot))  # <class 'bool'>
```

Конструкція `int(input(...))` працює «зсередини назовні»: спочатку `input()` отримує рядок, потім `int()` перетворює його на ціле число.

Назви булевих змінних часто починають з `is_` або `has_` — так одразу видно, що там «так/ні».

> 💡 **Трохи теорії**

Число **парне**, якщо ділиться на 2 без остачі, тобто остача `% 2` дорівнює 0:

```python
print(8 % 2)    # 0 → парне
print(7 % 2)    # 1 → непарне
```

Вираз `number % 2 == 0` спершу обчислює остачу, а потім порівнює її з нулем — і дає `True` або `False`.

Так само можна перевіряти подільність на будь-яке число: `n % 3 == 0` — ділиться на 3.

> 💡 **Трохи теорії**

**Логічні оператори** поєднують кілька умов:

| Оператор | Результат `True`, коли... | Приклад |
|---|---|---|
| `and` | **обидві** умови істинні | `True and False` → `False` |
| `or` | **хоча б одна** умова істинна | `True or False` → `True` |
| `not` | перевертає значення | `not True` → `False` |

```python
age = 15
print(age > 12 and age < 20)   # True — підліток
```

> 💡 **Трохи теорії**

Знадобиться `if` / `else` (див. теорію до Завдання 20):

```python
answer = input("Любиш каву? ")
if answer == "так":
    print("Я теж!")
else:
    print("Може, чай?")
```

Порівняння рядків **чутливе до регістру**: `"python" == "Python"` → `False`. Щоб не залежати від того, як користувач написав слово, приведи його до одного регістру:

```python
lang = input("Улюблена мова: ").strip().lower()   # strip() прибирає зайві пробіли
if lang == "python":
    ...
```

Ім'я користувача можна гарно вставити у відповідь за допомогою f-рядка.
